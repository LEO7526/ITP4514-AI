```python
from constraint import *  # Importing the CSP (Constraint Satisfaction Problem) library

# Magic Square with N Using CSP

n = 4  # Define the value of n (the size of the magic square)

# Calculate the magic sum for a magic square of size n x n
# The formula for the magic sum is n * (n^2 + 1) // 2
csum = n * (n * n + 1) // 2

# Define the indices for the left diagonal and right diagonal
ldiag = [(n + 1) * i for i in range(n)]  # [(3 + 1) * 0, (3 + 1) * 1, (3 + 1) * 2] , [0, 4, 8] → The indices for the left diagonal.
rdiag = [(n - 1) * (i + 1) for i in range(n)]  # [(3 - 1) * (1), (3 - 1) * (2), (3 - 1) * (3)], [2, 4, 6] → The indices for the right diagonal.

# Create a CSP problem instance
problem = Problem()

# Add variables for each cell in the n x n magic square
# Each cell can take a value from 1 to n^2
problem.addVariables(range(0, n * n), range(1, n * n + 1))

# Add a constraint that all values must be different
problem.addConstraint(AllDifferentConstraint(), range(0, n * n))

# Add a constraint for the left diagonal to have the sum equal to the magic sum
problem.addConstraint(ExactSumConstraint(csum), ldiag)

# Add a constraint for the right diagonal to have the sum equal to the magic sum
problem.addConstraint(ExactSumConstraint(csum), rdiag)

# Add a constraint for each row to have the sum equal to the magic sum
for row in range(n):
    # The indices for each row are [row*n + i for i in range(n)]
    problem.addConstraint(ExactSumConstraint(csum), [row * n + i for i in range(n)])

# Add a constraint for each column to have the sum equal to the magic sum
for col in range(n):
    # The indices for each column are [col + n*i for i in range(n)]
    problem.addConstraint(ExactSumConstraint(csum), [col + n * i for i in range(n)])

# Get a solution to the problem (returns a dictionary of variable assignments)
solution = problem.getSolution()

# If a solution is found, print it in a formatted n x n grid
if solution:
    print("Magic Square ({}x{}):".format(n, n))
    for row in range(n):
        # Print each row of the magic square using the solution dictionary
        print([solution[row * n + i] for i in range(n)])
else:
    print("No solution found.")
```

### Ex 2 - N-Rook Problem

```python
from constraint import *  # Importing the required library for constraint satisfaction

n = 5  # Number of rooks and size of the chessboard (n x n)

cols = range(1, n+1)  # Columns are treated as variables, ranging from 1 to n
rows = range(1, n+1)  # Rows are treated as domains for each column, ranging from 1 to n

problem = Problem()  # Create a new problem instance for CSP
problem.addVariables(cols, rows)  # Add variables to the problem with the given domains

# Goal: Assign a value (row) to each column such that no two columns share the same row

for col1 in cols:
    for col2 in cols:
        if col1 < col2:      # Add a constraint that ensures no two rooks are placed in the same row
            problem.addConstraint(lambda row1, row2, col1=col1, col2=col2:
                row1 != row2, (col1, col2))

# Get all possible solutions to the problem
solutions = problem.getSolutions()

# Print each solution found
for sol in solutions:
    print(sol)

# Print the total number of solutions found
print('Solutions found : %i' % len(solutions))
```

```python
from constraint import Problem, AllDifferentConstraint

n = 5
cols = range(1, n+1)  # Variables: columns 1 to n
rows = range(1, n+1)  # Domains: rows 1 to n

problem = Problem()
problem.addVariables(cols, rows)

# Enforce that all rooks are in different rows
problem.addConstraint(AllDifferentConstraint(), cols)

solutions = problem.getSolutions()

# Function to print the chessboard for a given solution
def print_board(solution):
    # Initialize an empty n x n board represented by '.'
    board = [['.' for _ in range(n)] for _ in range(n)]
    # Place 'R' where the rooks are located
    for col, row in solution.items():
        # Adjust indices since lists are 0-based
        board[row - 1][col - 1] = 'R'
    # Print the board row by row
    for row in board:
        print(' '.join(row))
    print()  # Add an empty line after each board

# Iterate over solutions and print each in chessboard format
for idx, sol in enumerate(solutions, 1):
    print(f"Solution #{idx}:")
    print_board(sol)
print('Solutions found : %i' % len(solutions))
```

### Ex 3 - Factorial

```python
n = var()                 # Create a logic variable 'n' using kanren's var() function
                          # n is now a logic variable that can be unified with values during pattern matching.

# Define the factorial function using pattern matching
@match(0)                 # Base Case Function: @match(0) decorator call this function when the argument n matches the pattern 0.
def factl(n):             # When n is 0, the factorial is defined as 1.
    return 1              # This function returns 1 for the base case of the recursion.

@match(n)                 # Recursive Case Function: @match(n) decorator is intended to match any value of n
def factl(n):             # Recursive Calls: factl(3), factl(2), factl(1) and factl(0).
    return n * factl(n-1) # For any n not equal to 0, this function computes n * factl(n - 1).

""" Flow of Execution for factl(4)
    factl(4)
    ├── 4 * factl(3)
         ├── 3 * factl(2)
              ├── 2 * factl(1)
                   ├── 1 * factl(0)
                        ├── returns 1 (base case)  Termination of Recursion
                   ├── returns 1 * 1 = 1           Unwinding the Recursion
              ├── returns 2 * 1 = 2
         ├── returns 3 * 2 = 6
    ├── returns 4 * 6 = 24
Base Case: Stops the recursion and provides a direct answer.
Recursive Case: Breaks the problem into smaller parts and calls the function itself.
"""
factl(5)
```

```python
# Defining relationships
parent = Relation()
facts(parent, ('Tom', 'Bobby'),
              ('Mary', 'Bobby'),
              ('Tom', 'Amy'),
              ('Mary', 'Amy'))

# Defining male and female relationships
male = Relation()
fact(male, 'Tom')
fact(male, 'Bobby')

female = Relation()
fact(female, 'Mary')
fact(female, 'Amy')

# Defining father relationship
def father(x, y):
    return lall(parent(x, y), male(x))

# Defining mother relationship
def mother(x, y):
    return lall(parent(x, y), female(x))

# Revised brother relationship
def brother(x, y):
    z = var()  # Common parent variable
    return lall(
        parent(z, x),  # z is a parent of x
        parent(z, y),  # z is a parent of y
        male(x),       # x must be male
        neq(x, y)      # x and y must be different individuals
    )

# Defining sister relationship
def sister(x, y):
    z = var()  # Common parent variable
    return lall(
        parent(z, x),  # z is a parent of x
        parent(z, y),  # z is a parent of y
        female(x),     # x must be female
        neq(x, y)      # x and y must be different individuals
    )

# Queries for relationships
# Find all children of 'Tom'
x = var()
print("Children of Tom:", run(0, x, father('Tom', x)))  # Expected output: ('Bobby', 'Amy')

# Find all children of 'Mary'
x = var()
print("Children of Mary:", run(0, x, mother('Mary', x)))  # Expected output: ('Bobby', 'Amy')

# Find brothers of 'Bobby'
x = var()
result = run(0, x, brother('Amy', x))
print("Brothers of Bobby:", result)
```
